# **Endless Racing Car Game!**

```python
# Importing Modules
import pygame
import random
import sys
pygame.init()

# Making the Display
w = 1370
h = 705
screen = pygame.display.set_mode((w, h))
pygame.display.set_caption("Endless Racing Game!")
clock = pygame.time.Clock()

# Uploading Road and Music Image
road = pygame.image.load("road.png")
road = pygame.transform.scale(road, (w, h))

pygame.mixer.music.load("music.mp3")
pygame.mixer.music.play(-1)

# Uploading Images of Cars
mycar = pygame.image.load("car.png")
policecar = pygame.image.load("enemy_car.png")
badcars = [
    pygame.image.load("ddf.png"),
    pygame.image.load("dhd.png"),
    pygame.image.load("fdf.png"),
    pygame.image.load("xuc.png"),
    pygame.image.load("xyz.png"),
    pygame.image.load("zyz.png")
]

# Making All Cars Small
mycar = pygame.transform.scale(mycar, (50, 100))
policecar = pygame.transform.scale(policecar, (50, 100))
badcars = [pygame.transform.scale(x, (50, 100)) for x in badcars]

# Adding Road Boundaries
leftroad = 250
rightroad = 1120

# Determining The Place of The Car
myx = (leftroad + rightroad)//2
myy = h - 120
myspeed = 6

# Telling Enemies and Gap
enemys = []
gap = 150

# Creating Score Calculator
font = pygame.font.Font(None, 40)
score = 0

# Generating Random Enemies
def make_enemy():
    if random.randint(1, 8) == 1:
        img = policecar
    else:
        img = random.choice(badcars)

    for i in range(50):
        x = random.randint(leftroad, rightroad - 50)
        y = random.randint(-300, -100)
        new_car_position = (x, y)
        new_car_size = (50, 100)
        
        crash = False
        for e in enemys:
            ex, ey = e[1], e[2]
            if (x < ex + 50 and x + 50 > ex and y < ey + 100 + gap and y + 100 > ey):
                crash = True
                break

        if not crash:
            enemys.append([img, x, y])
            break

# Score Showing
def show_score():
    screen.fill((0, 0, 0))
    text = font.render("Score: " + str(score), True, (255, 255, 255))
    screen.blit(text, (w//2 - text.get_width()//2, h//2))
    pygame.display.flip()

# Continuation of The Game
def game():
    global myx, score, enemys
    enemys = []
    myx = (leftroad + rightroad)//2
    run = True

    while run:
        screen.blit(road, (0, 0))

        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()

        keys = pygame.key.get_pressed()
        if keys[pygame.K_LEFT] and myx > leftroad:
            myx -= myspeed
        if keys[pygame.K_RIGHT] and myx < rightroad - 50:
            myx += myspeed

        if random.randint(1, 30) == 1:
            make_enemy()

        espeed = 5 + score // 10

        for e in enemys:
            e[2] += espeed
            screen.blit(e[0], (e[1], e[2]))

            # Checking for a crash!
            if (myx < e[1] + 50 and myx + 50 > e[1] and myy < e[2] + 100 and myy + 100 > e[2]):
                run = False

        newlist = []
        for e in enemys:
            if e[2] >= h:
                score += 1
            else:
                newlist.append(e)
        enemys = newlist

        screen.blit(mycar, (myx, myy))

        text = font.render("Score: " + str(score), True, (255, 255, 255))
        screen.blit(text, (10, 10))

        pygame.display.flip()
        clock.tick(60)

def wait():
    show_score()
    start = pygame.time.get_ticks()
    while pygame.time.get_ticks() - start < 10000:
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()
        pygame.time.delay(100)

# Loading The Game Forever
while True:
    game()
    wait()
    score = 0
```
